# Cálculo de resistencia térmica de growler en agua

```python
import numpy as np
import matplotlib.pyplot as plt
import CoolProp.CoolProp as CP
```

```python
def Grashof(beta, T_inf, T_s, L_c, nu):
        return (9.81 * beta * (T_inf - T_s) * (L_c**3)) / (nu**2)

def conveccion_natural_placa_vertical(Gr, Pr, k, L_c):
        Ra = Pr * Gr
        denominador = 1 + ((0.492 / Pr) ** (9/16))
        numerador = 0.287 * (Ra ** (1/6))
        Nu = (0.825 + (numerador / (denominador ** (8/27)))) ** 2
        h = Nu * k / L_c
        return h
```

Definimos parámetros

```python
T_water_in = 18 + 273.15  # Temperatura del agua interior en Kelvin
T_water_out = 10 + 273.15  # Temperatura del agua exterior en Kelvin
e_botella = 0.01  # Espesor de la botella en metros
k_botella = 0.2  # Conductividad térmica de la botella en W/mK
L_c = 0.3  # Altura de la botella en metros
D_botella = 0.1  # Diámetro de la botella en metros
Area_superficial_balde = np.pi * D_botella * L_c  # Área superficial del balde
```

```python
def calcular_h_vertical_agua(T):
    rho_water = CP.PropsSI('D', 'T', T, 'P', 101325, 'Water') 
    k_water = CP.PropsSI('L', 'T', T, 'P', 101325, 'Water') 
    beta_water = CP.PropsSI('ISOBARIC_EXPANSION_COEFFICIENT', 'T', T, 'P', 101325, 'Water')
    cp_water = CP.PropsSI('C', 'T', T, 'P', 101325, 'Water')
    mu_water = CP.PropsSI('V', 'T', T, 'P', 101325, 'Water')
    nu_water = mu_water / rho_water
    Pr = cp_water * mu_water / k_water
    Gr = Grashof(beta_water, T_water_in, T_water_out, L_c, nu_water)
    h_conv_natural = conveccion_natural_placa_vertical(Gr, Pr, k_water, L_c)
    return h_conv_natural

h_exterior = calcular_h_vertical_agua(T_water_out)
```
