```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
k = np.array([1.2, 2.3, 3.5, 4.6, 5.7])
h = np.array([1.4, 2.1, 3.0, 3.8, 4.7])

# Punto seleccionado
k_sel = 5.7
h_sel = 4.7

# Figura
fig, ax = plt.subplots(figsize=(10, 7), dpi=600)
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# Límites
ax.set_xlim(0, 6.5)
ax.set_ylim(0, 5.7)

# Quitar ejes normales
ax.axis("off")

# Ejes con flechas
ax.arrow(
    0.6, 0.6, 5.5, 0,
    head_width=0.18, head_length=0.22,
    fc="black", ec="black",
    linewidth=4,
    length_includes_head=True
)

ax.arrow(
    0.6, 0.6, 0, 4.9,
    head_width=0.18, head_length=0.22,
    fc="black", ec="black",
    linewidth=4,
    length_includes_head=True
)

# Línea punteada
ax.plot(
    k, h,
    linestyle=(0, (5, 5)),
    color="0.35",
    linewidth=2.5,
    zorder=1
)

# Puntos azules
ax.scatter(
    k[:-1], h[:-1],
    s=260,
    color="#0057e7",
    edgecolor="#0057e7",
    zorder=3
)

# Punto seleccionado como círculo rojo con X
ax.scatter(
    [k_sel], [h_sel],
    s=420,
    facecolors="white",
    edgecolors="red",
    linewidths=3,
    zorder=4
)

ax.scatter(
    [k_sel], [h_sel],
    s=280,
    marker="x",
    color="red",
    linewidths=3,
    zorder=5
)

# Texto selected
ax.text(
    5.45, 5.55,
    "selected",
    fontsize=30,
    fontweight="bold",
    ha="center",
    va="bottom"
)

# Flecha hacia el punto seleccionado
ax.annotate(
    "",
    xy=(k_sel + 0.18, h_sel + 0.18),
    xytext=(6.25, 5.35),
    arrowprops=dict(
        arrowstyle="-|>",
        color="black",
        linewidth=2.8,
        mutation_scale=24
    )
)

# Etiquetas de ejes
ax.text(
    3.3, 0.05,
    r"$k$",
    fontsize=30,
    ha="center",
    va="center"
)

ax.text(
    0.05, 3.0,
    r"$\hat{h}_k$",
    fontsize=30,
    ha="center",
    va="center"
)

plt.tight_layout()
plt.savefig("step1_generated.png", dpi=300, bbox_inches="tight", facecolor="white")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos para la función escalonada acumulada
x_edges = np.array([0.0, 0.8, 1.9, 3.0, 4.1, 5.2, 6.5, 8.1, 9.8])
y_vals  = np.array([0.01, 0.10, 0.24, 0.43, 0.60, 0.76, 0.92, 0.98])

# Figura
fig, ax = plt.subplots(figsize=(10, 7), dpi=600)
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# Quitar ejes por defecto
ax.axis("off")

# Límites
ax.set_xlim(-0.8, 10.5)
ax.set_ylim(-0.12, 1.18)

# Ejes con flecha
ax.arrow(
    0, 0, 10.1, 0,
    head_width=0.03, head_length=0.15,
    fc="black", ec="black",
    linewidth=2.5,
    length_includes_head=True
)

ax.arrow(
    0, 0, 0, 1.12,
    head_width=0.15, head_length=0.03,
    fc="black", ec="black",
    linewidth=2.5,
    length_includes_head=True
)

# Área rellena de la acumulada
ax.stairs(
    y_vals, x_edges,
    fill=True,
    baseline=0,
    facecolor="#f28b82",
    edgecolor="black",
    linewidth=2.0,
    alpha=0.9
)

# Etiquetas de ticks en y
ax.plot([-0.12, 0], [0, 0], color="black", lw=2)
ax.plot([-0.12, 0], [1, 1], color="black", lw=2)
ax.text(-0.25, 0, "0", fontsize=30, ha="right", va="center")
ax.text(-0.25, 1, "1", fontsize=30, ha="right", va="center")

# Etiquetas de ejes
ax.text(5.1, -0.08, r"$\hat{h}_k$", fontsize=30, ha="center", va="top")
ax.text(-0.9, 0.55, "Cumulative probability", fontsize=30, rotation=90,
        ha="center", va="center")

# Guardar y mostrar
plt.tight_layout()
plt.savefig("cumulative_probability_generated.png", dpi=300,
            bbox_inches="tight", facecolor="white")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Datos experimentales
# =========================

# Alturas: h8 = 20 cm, h7 = 40 cm, ..., h1 = 160 cm
H_cm = np.array([160, 140, 120, 100, 80, 60, 40, 20])
H_m = H_cm / 100

# Tiempos medidos en segundos para llenar 200 mL
tiempos = np.array([
    [12.79, 13.42, 13.86],  # h1
    [11.38, 12.27, 12.19],  # h2
    [10.81, 11.89, 12.00],  # h3
    [9.99,  9.86,  10.11],  # h4
    [9.53,  9.80,  10.04],  # h5
    [8.80,  8.86,  9.26],   # h6
    [8.60,  8.63,  9.10],   # h7
    [8.60,  8.30,  8.60]    # h8
])

# Volumen llenado
V_ml = 200
V_L = V_ml / 1000  # litros

# =========================
# Cálculo de caudal
# =========================

t_prom = np.mean(tiempos, axis=1)
t_std = np.std(tiempos, axis=1, ddof=1)

# Caudal en L/s
Q_Ls = V_L / t_prom

# Caudal en L/min
Q_Lmin = Q_Ls * 60

# Incertidumbre aproximada del caudal
Q_std_Lmin = Q_Lmin * (t_std / t_prom)

# =========================
# Gráfico H vs Q
# =========================

plt.figure(figsize=(7, 5), dpi=300)

plt.errorbar(
    Q_Lmin,
    H_m,
    xerr=Q_std_Lmin,
    fmt='o',
    capsize=4,
    label='Datos experimentales'
)

plt.plot(Q_Lmin, H_m, linestyle='--')

plt.xlabel('Caudal Q [L/min]', fontsize=12)
plt.ylabel('Altura H [m]', fontsize=12)
plt.title('Curva experimental de la bomba H vs Q', fontsize=13)

plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

# =========================
# Mostrar tabla de resultados
# =========================

for i in range(len(H_m)):
    print(
        f"h{i+1}: H = {H_m[i]:.2f} m, "
        f"t_prom = {t_prom[i]:.2f} s, "
        f"Q = {Q_Lmin[i]:.3f} L/min"
    )
```
